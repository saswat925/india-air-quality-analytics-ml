```python


START_DATE = "2026-08-01"
END_DATE = "2026-09-14"

hourly_variables = [
    "pm2_5",
    "pm10",
    "carbon_monoxide",
    "nitrogen_dioxide",
    "sulphur_dioxide",
    "ozone",
    "us_aqi"
]

print(hourly_variables)
```

```python
city = cities[0]

url = "https://air-quality-api.open-meteo.com/v1/air-quality"

params = {
    "latitude": city["lat"],
    "longitude": city["lon"],
    "hourly": ",".join(hourly_variables),
    "start_date": START_DATE,
    "end_date": END_DATE,
    "timezone": "Asia/Kolkata"
}

response = requests.get(url, params=params, timeout=60)

print("Status Code:", response.status_code)
```

```python
data = response.json()

print("Available keys:")
print(data["hourly"].keys())

print("\nNumber of records:")
print(len(data["hourly"]["time"]))
```

```python
def fetch_aqi_data(city_info, start_date, end_date):
    
    url = "https://air-quality-api.open-meteo.com/v1/air-quality"
    
    params = {
        "latitude": city_info["lat"],
        "longitude": city_info["lon"],
        "hourly": ",".join(hourly_variables),
        "start_date": start_date,
        "end_date": end_date,
        "timezone": "Asia/Kolkata"
    }
    
    response = requests.get(url, params=params, timeout=60)
    
    # API response check
    response.raise_for_status()
    
    data = response.json()
    
    # Convert hourly JSON to DataFrame
    df = pd.DataFrame(data["hourly"])
    
    # Add city information
    df["city"] = city_info["city"]
    df["state"] = city_info["state"]
    df["latitude"] = city_info["lat"]
    df["longitude"] = city_info["lon"]
    
    return df
```

```python
test_df = fetch_aqi_data(
    cities[0],
    START_DATE,
    END_DATE
)

print("Shape:", test_df.shape)
print(test_df.head())
```

```python
all_data = []

for i, city in enumerate(cities, start=1):
    
    print(f"Fetching {i}/{len(cities)}: {city['city']}")
    
    try:
        df_city = fetch_aqi_data(
            city,
            START_DATE,
            END_DATE
        )
        
        all_data.append(df_city)
        
        print(f"   Rows fetched: {len(df_city)}")
        
    except Exception as e:
        print(f"   ERROR: {e}")
    
    time.sleep(1)
```

```python
aqi_raw = pd.concat(
    all_data,
    ignore_index=True
)

print("Final shape:", aqi_raw.shape)
```

Create a cleaned working dataset

```python
# Create a separate cleaned dataset
aqi_clean = aqi_raw.copy()

# Extract date/time features
aqi_clean["date"] = aqi_clean["time"].dt.date
aqi_clean["year"] = aqi_clean["time"].dt.year
aqi_clean["month"] = aqi_clean["time"].dt.month
aqi_clean["day"] = aqi_clean["time"].dt.day
aqi_clean["hour"] = aqi_clean["time"].dt.hour
aqi_clean["day_of_week"] = aqi_clean["time"].dt.dayofweek

# Create readable day name
aqi_clean["day_name"] = aqi_clean["time"].dt.day_name()

print("Shape:", aqi_clean.shape)

print("\nNew columns:")
print(aqi_clean.columns.tolist())

print("\nSample:")
display(aqi_clean.head())
```

```python
##Check the new features
print("Hour range:", aqi_clean["hour"].min(), "to", aqi_clean["hour"].max())
print("Month values:", sorted(aqi_clean["month"].unique()))
print("Day of week values:", sorted(aqi_clean["day_of_week"].unique()))
print("Day names:", aqi_clean["day_name"].unique())
```

Create AQI Category
Now we add a business-friendly target category from us_aqi.
This is useful for EDA, visualization, and later classification ML.

```python
def classify_aqi(aqi):
    if aqi <= 50:
        return "Good"
    elif aqi <= 100:
        return "Moderate"
    elif aqi <= 150:
        return "Unhealthy for Sensitive Groups"
    elif aqi <= 200:
        return "Unhealthy"
    elif aqi <= 300:
        return "Very Unhealthy"
    else:
        return "Hazardous"


aqi_clean["aqi_category"] = aqi_clean["us_aqi"].apply(classify_aqi)

print(aqi_clean[["us_aqi", "aqi_category"]].head(20))
```

```python
##Then check the distribution:
print("\nAQI Category Distribution:")
print(aqi_clean["aqi_category"].value_counts())

print("\nAQI Category Percentage:")
print(
    (aqi_clean["aqi_category"].value_counts(normalize=True) * 100)
    .round(2)
)
```
